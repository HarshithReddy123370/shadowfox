```python
plt.figure(figsize=(15,5))
smoker_numeric = pd.get_dummies(df['smoker'], drop_first=True)
smoker_plot = sns.lineplot(data=smoker_numeric.rolling(20).mean())

plt.title("Smoker")
plt.ylabel("no of people")
plt.xlabel("Time")
h,_=smoker_plot.get_legend_handles_labels()
smoker_plot.legend(h,["Smoker"])
```

```python
plt.figure(figsize=(15,5))
children_numeric = pd.get_dummies(df['children'], drop_first=True)
children_plot = sns.lineplot(data=children_numeric.rolling(20).mean())
plt.title("Children")
plt.ylabel("no of people")
plt.xlabel("Time")
h,_=children_plot.get_legend_handles_labels()
children_plot.legend(h,["Children"])
```
